```python
def anagram_string(string1:str , string2:str)-> bool:
    """ check whether two string are anagram using character counting
    parameter: 
            take two string 
    Return:
        bool: true if string are anagram otherwise false
    Reises :
        TypeError: if input in not a string
    """
    if not isinstance(string1,str) or not isinstance(string2,str):
        raise TypeError("Input must be a String")
    char_count={}
    char_count1={}
    if len(string1)!=len(string2):
        return False
    else:
        for item in string1.lower().replace(" ",""):
            if item in char_count:
                char_count[item]+=1
            else:
                char_count[item]=1
        # print(char_count)
        for item in string2.lower().replace(" ",""):
            if item in char_count1:
                char_count1[item]+=1
            else:
                char_count1[item]=1
        # print(char_count1)
        if char_count== char_count1:
            return True
        else:
            return False
if __name__ =="__main__":
    try:
        user_input1="silent"
        user_input2="listen"
        result = anagram_string(user_input1,user_input2)
        print(result)
    except Exception as e:
        print(e)
```
