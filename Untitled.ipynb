```python
"""
write a function to check if two input string are anagrams of each other 

ex: str1 = "listen", str2="silent"  
means if we suffel the character there occurance must be same
"""
```

```python
string1="sakhsam"
string2='saksham'
char_count = {}
if len(string1)!=len(string2):
    print(False)
else:
    for item in string1.lower():
        if item in char_count:
            char_count[item]+=1
        else:
            char_count=1
    print(char_count)
```
