```python
def rev_sentence(user_input:str)->str:
    """
    Reverse the order of word in a sentence

    Args: 
       sentence(str) : input sentence
    Return:
    str: sentence 
    """
    if not isinstance(sentence,str):
        raise TypeError("input must be string")
    char_list=[]
    char=""
    for item in user_input:
        if item !=" ":
            char=char+item
        else:
            char_list.append(char)
            char=""
    char_list.append(char)
    
    rev_str=""
    for i in range(len(char_list)-1,-1,-1):
        rev_str=rev_str+char_list[i]
        if i!=0:
            rev_str= rev_str+" "
    return rev_str

if __name__=="__main__":
    my_sentence="This is python program"
    try:
        result=rev_sentence(my_sentence)
        print(result)
    except Exception as e:
        print(e)
```
