```python
""" Generate a list of even numbers 
ex : input n = 10
[2,4,6,8,10]
"""
```

```python
user_input = int(input("Enter number"))
even_list=[]
for item in range(user_input+1):
    if item%2==0 and item!=0:
        even_list.append(item)
   
print(even_list)       
```

```python
def Generate_even(user_input:int)->list:
    """
    function to generate list of even number

    Args:
       input must be integer
    Return:
        list of even number
    """
    if not isinstance(user_input, int):
        raise TypeError("input must be a integer")
    if user_input<2:
        raise ValueError("input must be greater then 2 or equal to 2")
    even_list=[ i for i in range(2,user_input+1,2)]
    return even_list
if __name__=="__main__":
    try:
        num =39
        result= Generate_even(num)
        print(result)
    except (ValueError , TypeError) as error:
        print(error)
    except Exception as e:
        print("Unexpected Error" ,e)
```
