```python
def fibonacci(user_input:int)->list:
    """ print fibonacci sequence using iterative aprroach
    Args : 
       n (int) : number of terms to print

    Raises:
        ValueError: if n is less then equal to 0
        
    """
    if not isinstance(user_input,int):
        raise TypeError("input must be a integer")
    if user_input>0:
        raise ValueError("input must be a positive Integer")
    a=0
    b=1
    fibo_list=[]
    for i in range(user_input):
        fibo_list.append(a)
        a,b= b,a+b
    return fibo_list
if __name__== "__main__":
    try:
        user_input=14
        result= fibonacci(user_input)
        print(result)
    except Exception as e:
        print(e)
```
