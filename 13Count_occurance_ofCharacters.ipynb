```python
string="i am a software developer in google"
char_count={}
for item in string.lower().replace(" ",""):
    if item in char_count:
        char_count[item]+=1
    else:
        char_count[item]=1
print(char_count)
```

```python
def count_character(string:str)->dict:
    """
    Function to count the occurences of each character in a string

    Args:
    string : the string whose characters are to be counted

    Returns: 
          dict:  a dictionary with characters as keys and their counts as values
    Raises: 
        TypeError: if input is not a string
    """
    if not isinstance(string,str):
        raise TypeError("input must be a string")
    char_count={}
    for item in string.lower().replace(" ",""):
        if item in char_count:
            char_count[item]+=1
        else:
            char_count[item]=1
    return char_count

if __name__=="__main__":
    user_input = input("Enter the string")
    try:
        result = count_character(user_input)
        print(result)
    except TypeError as e:
        print("Error",e)
    except Exception as e:
        print(e)
```
