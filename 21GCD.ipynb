```python
""" Write a function to find the greatest common divisor(GCD) of two numbers 
ex:  input  a =48 , b=18
output 6(the gcd of 48 18 is 6) 
"""
```

```python
a =48
b= 18
a_list=[]
b_list=[]
gcd=0
for i in range(2,a+1):
    if a%i==0:
        a_list.append(i)
for i in range(2,b+1):
    if b%i==0:
        b_list.append(i)
for i in range(len(b_list) - 1, -1, -1):
    if b_list[i] in a_list:
        gcd = b_list[i]
        break
        
print(a_list)
print(b_list)
print("GCD : " , gcd)
```

```python
def gcd_finder(user_input1:int , user_input2: int)-> int:
    """
    Find GCD of two numbers

    Args: 
       Takes two integer 

    Return:
        Gives an integer value 
    """
    if not isinstance(user_input1 and user_input2,int):
        raise TypeError("Input must be an integer value")
    small = min(user_input1,user_input2)
    gcd=0
    for i in range(small,0,-1):
        if user_input1%i==0 and user_input2%i==0:
            gcd=i
            break
    return gcd

if __name__ == "__main__":
    try:
        user_input1 = 6
        user_input2 = 30
        result = gcd_finder(int(user_input1),int(user_input2))
        print(result)
    except TypeError as e : 
        print("Input must be integer value")
    except Exception as e:
        print(e)
    
```
