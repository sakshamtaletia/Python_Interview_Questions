```python
def power_founder(base_value:int ,power_value:int)->int:
    """
    find power of number without using ** operator
    Paramater: 
            Take 2 Integer values
    Return:
          An Integer value
    """
    if not isinstance(base_value and power_value, int):
        raise TypeError(" Both value must be integer")
    if power_value <0:
        raise ValueError(" Power value must be a positive number")
    result = 1
    for item in range(power_value):
        result*=base_value
    return result
if __name__=="__main__":
    try:
        base_value=-2
        power_value=3
        result = power_founder(base_value,power_value)
        print(result)
    except (ValueError , TypeError )as error:
        print(error)
    except Exception as e:
        print("unexcepted error" , e)
```
